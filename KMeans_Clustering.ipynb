```python
import seaborn as sns
from sklearn.datasets import make_blobs
```

```python
X,y=make_blobs(
    n_samples=1000,
    n_features=2,
    centers=4,random_state=42
)
```

```python
y
```

```python
#visualise
sns.scatterplot(x=X[:,0],y=X[:,1],c=y)
#y ko generate krana h by kmeans clustering 
```

```python
#kmeans
from sklearn.cluster import KMeans
```

```python
#greedy algo : hr step pe best decision lete h algo
K=4
kmeans=KMeans(
    n_clusters=K,
    
    random_state=42
)
```

```python
labels=kmeans.fit_predict(X)
#label=cluster no.
```

```python
labels
```

```python
sns.scatterplot(x=X[:,0],y=X[:,1],c=labels)
```

```python
#choose k value:elbow,silhouette score
#elbow method
import warnings
warnings.filterwarnings("ignore")
wcss=[]
for k in range(1,21):
    kmeans=KMeans(n_clusters=k)
    kmeans.fit_predict(X)
    wcss.append(kmeans.inertia_)
```

```python
sns.lineplot(x=range(1,21),y=wcss,marker='o')
```

```python
#silhouette score
from sklearn.metrics import silhouette_score
```

```python
ss=[]
for k in range(2,21):
    kmeans=KMeans(n_clusters=k)
    labels=kmeans.fit_predict(X)
    score=silhouette_score(X,labels)
    ss.append(score)
```

```python
#plot k and ss
sns.lineplot(x=range(2,21),y=ss,marker='o')
```

```python
X=iris.data
y=iris.target
```

```python
#visualise
sns.scatterplot(x=X[:,0],y=X[:,2],c=y)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)

```

```python
X_scaled
```

```python

#optional-dimesntionality reduction using PCA
from sklearn.decomposition import PCA
pca=PCA(n_components=2)
pca_data=pca.fit_transform(X_scaled)
```

```python
#elbow method
wcss=[]
for k in range(1,11):
    kmeans=KMeans(n_clusters=k)
    kmeans.fit(pca_data)
    wcss.append(kmeans.inertia_)
```

```python
sns.lineplot(x=range(1,11),y=wcss,marker='o')
```

```python
#kmeans
kmeans=KMeans(n_clusters=3,random_state=10)
labels=kmeans.fit_predict(pca_data)

sns.scatterplot(x=pca_data[:,0],y=pca_data[:,1],c=labels)
sns.scatterplot(x=kmeans.cluster_centers_[:,0],y=kmeans.cluster_centers_[:,1],marker="x",c="red",s=200)
sns.scatterplot(x=kmeans.cluster_centers_[:,0],y=kmeans.cluster_centers_[:,1],marker="X",c="red",s=200)

```
